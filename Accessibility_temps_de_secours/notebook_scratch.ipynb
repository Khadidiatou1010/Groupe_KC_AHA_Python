# Géometrie France

```python
# Voyons les couches de notre geopackage
couches = gpd.list_layers('Data/Geographie/BDC_5-1_GPKG_LAMB93_D028-ED2026-03-15.gpkg')
print(couches)
```

```python
# Nous prenons la couche qui nous interesse
gpk_communes = gpd.read_file('Data/Geographie/BDC_5-1_GPKG_LAMB93_D028-ED2026-03-15.gpkg', layer = 'commune')
gpk_communes.head()
```

```python
# Prenons les routes aussi
gpk_routes = gpd.read_file('Data/Geographie/BDC_5-1_GPKG_LAMB93_D028-ED2026-03-15.gpkg', layer = 'troncon_de_route')
gpk_routes.head()
```

## Carte des routes et communes

```python
# On vérifie bien les systèmes de prj
gpk_communes.crs
```

```python
gpk_routes.crs
```

```python
gpk_communes = gpk_communes.to_crs(4326)
gpk_routes = gpk_routes.to_crs(4326)
```

```python
fig, ax = plt.subplots(figsize=(10, 10))

gpk_communes.plot(ax=ax, color="lightgrey", edgecolor="black", linewidth=0.5)

gpk_routes.plot(ax=ax, color="red", linewidth=1)

ax.set_axis_off()
plt.show()
```
